## Thực hiện phân tích dữ liệu bằng Python

#### Nhiệm vụ 1.1 – Import thư viện và đọc dữ liệu

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc dữ liệu
df = pd.read_csv("./data/penguins_lter.csv")

# Hiển thị 5 dòng đầu tiên
display(df.head())
```

#### Nhiệm vụ 1.2 – Khám phá dữ liệu

```python
# Kích thước dữ liệu
print("Kích thước:", df.shape)

# Tên các cột và kiểu dữ liệu
print(df.info())

# Thống kê mô tả các biến số
display(df.describe())

# Số lượng giá trị thiếu của từng cột
print(df.isnull().sum())
```

#### Nhiệm vụ 1.3 – Phân tích biến định lượng

```python
print(df.columns.tolist())
```

```python
# Lấy cột khối lượng và loại bỏ giá trị thiếu
mass = df["Body Mass (g)"].dropna()

print("Số lượng:", mass.count())
print("Trung bình:", mass.mean())
print("Trung vị:", mass.median())
print("Độ lệch chuẩn:", mass.std())
print("Nhỏ nhất:", mass.min())
print("Lớn nhất:", mass.max())

print("Tứ phân vị:")
print(mass.quantile([0.25, 0.5, 0.75]))
```

#### Nhiệm vụ 1.5 – Phân tích biến định tính

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histogram
sns.histplot(mass, bins=20, kde=True, ax=axes[0])
axes[0].set_title("Phân phối khối lượng chim cánh cụt")
axes[0].set_xlabel("Khối lượng (g)")
axes[0].set_ylabel("Tần suất")

# Boxplot
sns.boxplot(x=mass, ax=axes[1])
axes[1].set_title("Boxplot khối lượng chim cánh cụt")
axes[1].set_xlabel("Khối lượng (g)")

plt.tight_layout()
plt.show()
```

#### Nhiệm vụ 1.6 – Tính tỷ lệ phần trăm

```python
# Tính tỷ lệ từng loài theo phần trăm
species_percent = df["Species"].value_counts(normalize=True) * 100

print(species_percent.round(2))

# Vẽ biểu đồ tròn
species_percent.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6, 6)
)

plt.title("Tỷ lệ các loài chim cánh cụt")
plt.ylabel("")
plt.show()
```
